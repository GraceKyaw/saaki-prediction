# Temporal-feature FFNN training: SA-AKI sliding-window representation

The fourth of four training-setup notebooks.

Input is `data/sliding_window_representation.parquet`. Unlike the previous
three representations, this one holds several rows per patient: each row is one
non-overlapping 24-hour window, described by the first value, last value and
trend of each feature.

Multiple rows per patient changes three things relative to the other notebooks:

- Splitting is grouped by patient, so no patient has windows on both sides of
  the train/test boundary or across cross-validation folds.
- Class weights are computed from window-level label counts, which differ from
  the patient-level balance used elsewhere.
- Confidence intervals resample whole patients rather than individual rows.

The network architecture and search space are identical to the static FFNN
notebook, so the temporally-aware input is the only variable that differs
between the two models.

Two imbalance-handling conditions are run as complete, independent pipelines,
each with its own Optuna search, refit and test evaluation.

## Setup

Imports and file paths. The fit/transform and evaluation functions are imported
from `src/`. The cluster-aware bootstrap functions replace the row-level ones
used in the previous three notebooks.

```python
import json
import os
import sys
from pathlib import Path

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"

import numpy as np
import pandas as pd
import optuna
import keras
from sklearn.model_selection import train_test_split, StratifiedGroupKFold
from sklearn.metrics import (
    average_precision_score, roc_auc_score,
    recall_score, precision_score, f1_score,
)

sys.path.insert(0, str(Path.cwd().parent))
from src.preprocessing_fit_transform import (
    winsorize_fit, winsorize_transform,
    impute_fit, impute_transform,
    scale_fit, scale_transform,
    encode,
)
from src.evaluation_utils import (
    calibration_report, plot_calibration,
    subgroup_report,
    decision_curve, plot_decision_curve,
    cluster_bootstrap_ci, cluster_bootstrap_metrics_report,
    cluster_subgroup_report,
)

DATA_DIR = Path("../data")
DATA_PATH = DATA_DIR / "sliding_window_representation.parquet"

RANDOM_STATE = 42
N_TRIALS = 50
N_FOLDS = 5
EPOCH_CEILING = 200
EARLY_STOPPING_PATIENCE = 18
REFIT_VAL_FOLDS = 7   # StratifiedGroupKFold split for the early-stopping holdout

TARGET = "label"
GROUP_COL = "subject_id"
CONDITIONS = ["no imbalance handling", "class weighting"]

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 200)
optuna.logging.set_verbosity(optuna.logging.WARNING)

print("keras", keras.__version__, "| optuna", optuna.__version__)
```

## 1. Load the data

Prints shape and class balance at both window and patient level, and confirms
`subject_id` is present, which every split in this notebook depends on.

```python
df = pd.read_parquet(DATA_PATH)

print("shape:", df.shape)
print(f"windows: {len(df):,} | patients: {df[GROUP_COL].nunique():,} "
      f"| stays: {df['stay_id'].nunique():,}")
print(f"\nwindow-level positive rate:  {df[TARGET].mean() * 100:.2f}%")
print(f"patient-level positive rate: {df.groupby(GROUP_COL)[TARGET].max().mean() * 100:.2f}%")

windows_per_patient = df.groupby(GROUP_COL).size()
print(f"\nwindows per patient: mean {windows_per_patient.mean():.2f}, "
      f"median {int(windows_per_patient.median())}, max {windows_per_patient.max()}")
df.head()
```

```python
STAT_SUFFIXES = ("_first", "_last", "_trend")
FEATURE_COLS = [c for c in df.columns if c.endswith(STAT_SUFFIXES)] + ["admission_age"]
ID_COLS = ["stay_id", GROUP_COL, "window_id"]
CATEGORICAL = ["gender", "admission_type"]
DROP_FROM_X = ID_COLS + [TARGET]

REQUIRED = ID_COLS + CATEGORICAL + ["admission_age", TARGET]
missing = [c for c in REQUIRED if c not in df.columns]
unaccounted = sorted(set(df.columns) - set(FEATURE_COLS) - set(REQUIRED))

print(f"engineered feature columns: {len(FEATURE_COLS) - 1}  "
      f"(expected 22 features x 3 stats = 66)")
print(f"numeric columns for winsorize/impute/scale: {len(FEATURE_COLS)}")
print("missing required columns:", missing if missing else "none")
print("unaccounted columns:", unaccounted if unaccounted else "none")

assert not missing, f"missing required columns: {missing}"
assert not unaccounted, f"unexpected columns: {unaccounted}"
assert GROUP_COL in df.columns, "subject_id is required for patient-grouped splitting"
```

## 2. Patient-grouped train/test split

The split is made over unique patients, not rows. Each patient is labelled by
whether any of their windows is positive, and that per-patient label is used to
stratify an 80/20 split of the patient list. Every window then follows its
patient to the corresponding side.

```python
patient_labels = df.groupby(GROUP_COL)[TARGET].max()
patients = patient_labels.index.to_numpy()

train_patients, test_patients = train_test_split(
    patients, test_size=0.20, stratify=patient_labels.to_numpy(),
    random_state=RANDOM_STATE,
)

train_df = df[df[GROUP_COL].isin(set(train_patients))].reset_index(drop=True)
test_df = df[df[GROUP_COL].isin(set(test_patients))].reset_index(drop=True)

print(f"train: {len(train_df):,} windows from {train_df[GROUP_COL].nunique():,} patients, "
      f"window positive rate {train_df[TARGET].mean() * 100:.2f}%")
print(f"test:  {len(test_df):,} windows from {test_df[GROUP_COL].nunique():,} patients, "
      f"window positive rate {test_df[TARGET].mean() * 100:.2f}%")
```

```python
overlap = set(train_df[GROUP_COL]) & set(test_df[GROUP_COL])
print(f"patients appearing on both sides: {len(overlap)}")
assert not overlap, f"{len(overlap)} patients have windows in both train and test"

assert len(train_df) + len(test_df) == len(df), "windows lost or duplicated in the split"
print("every window is assigned exactly once, and no patient is split across sides")
```

## 3. Fold processing and class weights

Winsorize, imputation and scaling are fitted on the training portion passed to
this function and applied unchanged to the other portion. `encode()` is then
applied to both. The data is already a flat table, so no reshaping is needed.

Class weights follow `n_samples / (n_classes * n_samples_in_class)` computed
from window-level counts, recomputed wherever they are used. Window-level
balance differs from patient-level balance, so a value computed elsewhere is
never reused here.

```python
def process(fit_df, apply_df, feature_cols=FEATURE_COLS):
    """Fit Winsorize/impute/scale on fit_df, apply to both frames, then encode.

    Returns (fit_df_processed, apply_df_processed, fitted_params).
    """
    bounds = winsorize_fit(fit_df, feature_cols)
    fit_df = winsorize_transform(fit_df, bounds)
    apply_df = winsorize_transform(apply_df, bounds)

    medians = impute_fit(fit_df, feature_cols)
    fit_df = impute_transform(fit_df, medians)
    apply_df = impute_transform(apply_df, medians)

    scaler = scale_fit(fit_df, feature_cols)
    fit_df = scale_transform(fit_df, scaler, feature_cols)
    apply_df = scale_transform(apply_df, scaler, feature_cols)

    return encode(fit_df), encode(apply_df), {
        "bounds": bounds, "medians": medians, "scaler": scaler,
    }


def to_xy(processed_df):
    """Split a processed frame into a float32 model matrix and the target."""
    X = processed_df.drop(columns=DROP_FROM_X).to_numpy(dtype="float32")
    y = processed_df[TARGET].to_numpy(dtype="float32")
    return X, y


def class_weights(y):
    """Return n_samples / (n_classes * n_samples_in_class), from window counts."""
    counts = np.bincount(y.astype(int), minlength=2)
    n_samples, n_classes = len(y), 2
    return {c: float(n_samples / (n_classes * counts[c])) for c in (0, 1)}


print("window-level class weights on the full training split:")
print(" ", class_weights(train_df[TARGET].to_numpy()))
```

## 4. Model architecture

Identical to the static FFNN notebook: 1 to 3 hidden Dense layers with ReLU
activation and L2 weight regularization, each followed by Dropout, then a single
sigmoid output unit. Optimizer Adam, loss binary crossentropy, with AUPRC
tracked as a metric so early stopping can monitor it.

```python
def build_model(n_features, n_layers, units, dropout, l2_value, learning_rate):
    """Build and compile the feedforward network. Returns the compiled model."""
    model = keras.Sequential(name="temporal_ffnn")
    model.add(keras.layers.Input(shape=(n_features,)))

    for _ in range(n_layers):
        model.add(keras.layers.Dense(
            units, activation="relu",
            kernel_regularizer=keras.regularizers.l2(l2_value),
        ))
        if dropout > 0:
            model.add(keras.layers.Dropout(dropout))

    model.add(keras.layers.Dense(1, activation="sigmoid"))

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=learning_rate),
        loss="binary_crossentropy",
        metrics=[keras.metrics.AUC(curve="PR", name="auprc")],
    )
    return model


def early_stopping():
    """Return an EarlyStopping callback monitoring validation AUPRC."""
    return keras.callbacks.EarlyStopping(
        monitor="val_auprc", mode="max",
        patience=EARLY_STOPPING_PATIENCE,
        restore_best_weights=True,
    )
```

## 5. Optuna objective and search space

Hyperparameters are selected on inner folds and evaluated on a test set held
out of selection (Cawley & Talbot, 2010; Varma & Simon, 2006).

Folds come from `StratifiedGroupKFold` grouped by `subject_id`, so the
patient-grouping guarantee holds inside the search as well as at the
train/test boundary. Winsorize, imputation and scaling are fitted on each
fold's own training portion. For class weighting, the weight dictionary is
recomputed from that fold's window-level training labels. Training runs to a
ceiling of 200 epochs with early stopping on the validation fold, monitoring
AUPRC.

Mean fold AUPRC is reported after each fold, allowing a trial to be pruned
before all five folds run. The epoch reaching the best validation AUPRC is
stored on the trial for use at refit.

| Parameter | Status | Range / value |
|---|---|---|
| Number of hidden layers | Tuned | 1 to 3, integer |
| Units per layer | Tuned | 8 to 128, integer, log scale |
| Dropout rate | Tuned | 0.0 to 0.5 |
| L2 weight regularization | Tuned | 1e-6 to 1e-2, log scale |
| Learning rate | Tuned | 1e-4 to 1e-2, log scale |
| Batch size | Tuned | 16, 32, 64 or 128, categorical |
| Epochs | Fixed | Ceiling 200, early stopping |
| `class_weight` | Fixed | Computed from window counts, class weighting only |
| Activation | Fixed | ReLU |
| Optimizer | Fixed | Adam |

```python
def suggest_params(trial):
    return {
        "n_layers": trial.suggest_int("n_layers", 1, 3),
        "units": trial.suggest_int("units", 8, 128, log=True),
        "dropout": trial.suggest_float("dropout", 0.0, 0.5),
        "l2_value": trial.suggest_float("l2_value", 1e-6, 1e-2, log=True),
        "learning_rate": trial.suggest_float("learning_rate", 1e-4, 1e-2, log=True),
        "batch_size": trial.suggest_categorical("batch_size", [16, 32, 64, 128]),
    }


def make_objective(condition):
    """Return the Optuna objective for the given imbalance-handling condition."""

    def objective(trial):
        params = suggest_params(trial)
        batch_size = params.pop("batch_size")
        sgkf = StratifiedGroupKFold(n_splits=N_FOLDS, shuffle=True,
                                    random_state=RANDOM_STATE)
        splits = sgkf.split(train_df, train_df[TARGET], groups=train_df[GROUP_COL])

        fold_scores, fold_best_epochs = [], []
        for fold, (tr_idx, va_idx) in enumerate(splits):
            fold_train = train_df.iloc[tr_idx].reset_index(drop=True)
            fold_valid = train_df.iloc[va_idx].reset_index(drop=True)

            fold_train, fold_valid, _ = process(fold_train, fold_valid)
            X_tr, y_tr = to_xy(fold_train)
            X_va, y_va = to_xy(fold_valid)

            weights = class_weights(y_tr) if condition == "class weighting" else None

            keras.utils.set_random_seed(RANDOM_STATE)
            model = build_model(X_tr.shape[1], **params)
            history = model.fit(
                X_tr, y_tr,
                validation_data=(X_va, y_va),
                epochs=EPOCH_CEILING,
                batch_size=batch_size,
                class_weight=weights,
                callbacks=[early_stopping()],
                verbose=0,
            )

            proba = model.predict(X_va, verbose=0).ravel()
            fold_scores.append(average_precision_score(y_va, proba))
            fold_best_epochs.append(int(np.argmax(history.history["val_auprc"])) + 1)

            keras.backend.clear_session()

            trial.report(float(np.mean(fold_scores)), fold)
            if trial.should_prune():
                raise optuna.TrialPruned()

        trial.set_user_attr("mean_best_epoch", int(np.mean(fold_best_epochs)))
        return float(np.mean(fold_scores))

    return objective
```

## 6. Run the Optuna study, once per condition

50 trials per condition, maximising mean fold AUPRC, with median pruning
(Akiba et al., 2019). The two studies are independent.

```python
def run_study(condition):
    study = optuna.create_study(
        direction="maximize",
        sampler=optuna.samplers.TPESampler(seed=RANDOM_STATE),
        pruner=optuna.pruners.MedianPruner(n_startup_trials=5, n_warmup_steps=1),
        study_name=f"temporal_ffnn_{condition.lower().replace(' ', '_')}",
    )
    study.optimize(make_objective(condition), n_trials=N_TRIALS, show_progress_bar=True)

    pruned = len([t for t in study.trials if t.state == optuna.trial.TrialState.PRUNED])
    print(f"\n{condition}: {len(study.trials)} trials, {pruned} pruned")
    print(f"best mean CV AUPRC: {study.best_value:.4f}")
    print(f"best epoch (mean over folds): {study.best_trial.user_attrs['mean_best_epoch']}")
    print("best params:")
    for k, v in study.best_params.items():
        print(f"  {k}: {v}")
    return study
```

### No imbalance handling

```python
study_no_weighting = run_study("no imbalance handling")
```

### Class weighting

```python
study_class_weighting = run_study("class weighting")
```

## 7. Refit on the full training split

For each condition, Winsorize, imputation and scaling are fitted once on the
entire 80% training split. One model per condition is then trained on that
processed data using the condition's winning hyperparameters.

Early stopping needs a validation set. That holdout is also drawn with
`StratifiedGroupKFold`, so it stays grouped by patient like every other split
in this notebook. The class weight dictionary is computed from the full
training split's window labels. The test set takes no part in this step.

```python
def refit(study, condition):
    params = dict(study.best_params)
    batch_size = params.pop("batch_size")

    train_proc, test_proc, fitted = process(train_df.copy(), test_df.copy())
    X_train, y_train = to_xy(train_proc)

    # patient-grouped, stratified holdout for the early-stopping callback
    holdout = StratifiedGroupKFold(n_splits=REFIT_VAL_FOLDS, shuffle=True,
                                   random_state=RANDOM_STATE)
    fit_rows, val_rows = next(holdout.split(train_proc, train_proc[TARGET],
                                            groups=train_proc[GROUP_COL]))

    weights = class_weights(y_train) if condition == "class weighting" else None

    keras.utils.set_random_seed(RANDOM_STATE)
    model = build_model(X_train.shape[1], **params)
    history = model.fit(
        X_train[fit_rows], y_train[fit_rows],
        validation_data=(X_train[val_rows], y_train[val_rows]),
        epochs=EPOCH_CEILING,
        batch_size=batch_size,
        class_weight=weights,
        callbacks=[early_stopping()],
        verbose=0,
    )

    val_patients = set(train_proc.iloc[val_rows][GROUP_COL])
    fit_patients = set(train_proc.iloc[fit_rows][GROUP_COL])
    assert not (val_patients & fit_patients), "early-stopping holdout leaks patients"

    best_epoch = int(np.argmax(history.history["val_auprc"])) + 1
    print(f"{condition}: refit on {len(fit_rows):,} windows "
          f"({len(val_rows):,} held back for early stopping), best epoch {best_epoch}"
          + (f", class weights {weights}" if weights else ""))
    return model, test_proc, fitted, best_epoch


model_no_weighting, test_proc_no_weighting, fitted_no_weighting, epoch_no_weighting = refit(
    study_no_weighting, "no imbalance handling")
model_class_weighting, test_proc_class_weighting, fitted_class_weighting, epoch_class_weighting = refit(
    study_class_weighting, "class weighting")
```

## 8. Evaluate on the held-out test set

The Winsorize bounds, medians and scaler fitted in step 7 are applied to the
test set unchanged. Nothing is refitted on test. Metrics are computed per
window, in the usual way; clustering affects the confidence intervals in step
11, not the point estimates here.

```python
def evaluate(model, test_proc, condition):
    X_te, y_te = to_xy(test_proc)
    proba = model.predict(X_te, verbose=0).ravel()
    pred = (proba >= 0.5).astype(int)

    return proba, {
        "condition": condition,
        "AUPRC": average_precision_score(y_te, proba),
        "AUROC": roc_auc_score(y_te, proba),
        "recall_AKI": recall_score(y_te, pred),                    # positive class
        "specificity_non_AKI": recall_score(y_te, pred, pos_label=0),  # negative class
        "precision": precision_score(y_te, pred),
        "F1": f1_score(y_te, pred),
    }


proba_no_weighting, metrics_no_weighting = evaluate(
    model_no_weighting, test_proc_no_weighting, "no imbalance handling")
proba_class_weighting, metrics_class_weighting = evaluate(
    model_class_weighting, test_proc_class_weighting, "class weighting")

_, y_test = to_xy(test_proc_no_weighting)
test_patient_ids = test_proc_no_weighting[GROUP_COL].to_numpy()

print(f"test windows: {len(y_test):,} from {len(np.unique(test_patient_ids)):,} patients")
print(f"test window positive rate: {y_test.mean() * 100:.2f}%  "
      f"(AUPRC of a random baseline equals the positive rate)")
```

Calibration compares predicted probabilities against the rates observed at

## 11. Cluster-aware bootstrap confidence intervals

The previous three notebooks resampled individual rows, which treats each row
as an independent observation. That holds when there is one row per patient. It
does not hold here: a patient contributes several windows with correlated
physiology, so resampling rows independently understates the uncertainty.

`cluster_bootstrap_metrics_report` resamples whole patients instead, so all of
a patient's windows move together in each resample. The test set's
`subject_id` values are passed as `patient_ids`.

Predictions come from the models refitted in step 7. Only the test set is
resampled.

```python
from IPython.display import display

boot_no_weighting = cluster_bootstrap_metrics_report(
    y_test, proba_no_weighting, test_patient_ids, threshold=0.5,
    label="no imbalance handling")
boot_class_weighting = cluster_bootstrap_metrics_report(
    y_test, proba_class_weighting, test_patient_ids, threshold=0.5,
    label="class weighting")

display(boot_no_weighting)
display(boot_class_weighting)
```

## 12. Subgroup performance

Performance is reported separately by gender, age band and admission type. Each
subgroup uses the same cluster-aware intervals as step 11, with that group's
`subject_id` values passed as `patient_ids`, so the correction applies within
subgroups and not only to the whole test set.

Predictions come from the models refitted in step 7. Nothing is refitted or
retrained.

```python
# Subgroup labels come from the untransformed test windows: admission_age in
# test_proc_* is scaled, so age banding needs the raw values. Row order is
# unchanged, so these line up with y_test and the predicted probabilities.
gender_values = test_df["gender"]

age_bands = pd.cut(test_df["admission_age"], bins=[18, 40, 65, 90],
                   labels=["18-40", "40-65", "65-90"], include_lowest=True).astype(str)

# the encoded emergency/elective split; several raw admission_type categories
# are too small to bootstrap
admission_groups = test_proc_no_weighting["admission_type_elective"].map(
    {0: "Emergency", 1: "Elective"}
)

print("group sizes (windows, and distinct patients)")
for name, values in [("gender", gender_values), ("age band", age_bands),
                     ("admission type", admission_groups)]:
    counts = values.value_counts(dropna=False).sort_index().to_dict()
    patients = {g: test_df.loc[values == g, GROUP_COL].nunique() for g in counts}
    print(f"  {name}: windows {counts}")
    print(f"  {'':{len(name)}s}  patients {patients}")
print()
```

```python
SUBGROUPS = {
    "Gender": gender_values,
    "Age band": age_bands,
    "Admission type": admission_groups,
}
CONDITION_PROBA = {
    "no imbalance handling": proba_no_weighting,
    "class weighting": proba_class_weighting,
}

subgroup_tables = {}
for attribute, values in SUBGROUPS.items():
    for condition, proba in CONDITION_PROBA.items():
        key = f"{condition} -- {attribute}"
        subgroup_tables[key] = cluster_subgroup_report(
            y_test, proba, values, test_patient_ids, label=key)
```

## 13. Decision curve analysis

Net benefit is computed across the full range of thresholds and compared
against two reference strategies, flagging every window and flagging none
(Vickers & Elkin, 2006). A model is ahead over the range where its curve sits
above both reference lines. The 0.4 to 0.7 threshold range is printed alongside
the chart.

Predictions come from the models refitted in step 7. Nothing is refitted or
retrained.

```python
curve_no_weighting = decision_curve(y_test, proba_no_weighting)
curve_class_weighting = decision_curve(y_test, proba_class_weighting)

fig_dca = plot_decision_curve([(curve_no_weighting, "no imbalance handling"),
                               (curve_class_weighting, "class weighting")])

print(curve_no_weighting[(curve_no_weighting["threshold"] >= 0.4) & (curve_no_weighting["threshold"] <= 0.7)].round(4).to_string(index=False))
print()
print(curve_class_weighting[(curve_class_weighting["threshold"] >= 0.4) & (curve_class_weighting["threshold"] <= 0.7)].round(4).to_string(index=False))
```

## Output

Per condition: winning hyperparameters, test metrics, and the fitted Winsorize
bounds, medians and scaler parameters. Written to `data/`, which is gitignored.
The scaler is stored as its centre and scale arrays rather than the object
itself.

```python
def save_results(study, metrics, fitted, best_epoch, condition):
    payload = {
        "condition": condition,
        "best_params": study.best_params,
        "epoch_used_at_refit": best_epoch,
        "best_cv_auprc": study.best_value,
        "test_metrics": {k: v for k, v in metrics.items() if k != "condition"},
        "fitted_params": {
            "winsorize_bounds": {k: [float(lo), float(hi)] for k, (lo, hi) in fitted["bounds"].items()},
            "medians": {k: float(v) for k, v in fitted["medians"].items()},
            "scaler_center": [float(v) for v in fitted["scaler"].center_],
            "scaler_scale": [float(v) for v in fitted["scaler"].scale_],
            "scaler_feature_order": FEATURE_COLS,
        },
    }
    path = DATA_DIR / f"temporal_ffnn_{condition.lower().replace(' ', '_')}_results.json"
    path.write_text(json.dumps(payload, indent=2))
    print(f"wrote {path.name}  ({path.stat().st_size / 1024:.1f} KB)")


save_results(study_no_weighting, metrics_no_weighting, fitted_no_weighting,
             epoch_no_weighting, "no imbalance handling")
save_results(study_class_weighting, metrics_class_weighting, fitted_class_weighting,
             epoch_class_weighting, "class weighting")
```
